# leetcode 0015 三数之和 双指针方法

> 为什么没有用hash的方法：hash无法去重，唯一可以的方法是用set强制去重，但是还是需要On^2的复杂度

``` python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        n = len(nums)
        nums.sort()
        ans = []

        for x_idx in range(n):
            #提前剪枝非常聪明
            if nums[x_idx] > 0:
                break
            if x_idx > 0 and nums[x_idx] == nums[x_idx-1]:
                continue
            z_idx = n - 1
            target = -nums[x_idx]
            for y_idx in range(x_idx + 1, n):
                if y_idx > x_idx + 1 and nums[y_idx] == nums[y_idx - 1]:
                    continue
                while y_idx < z_idx and nums[y_idx] + nums[z_idx] > target:
                    z_idx -= 1
                if y_idx == z_idx:
                    break
                if nums[y_idx] + nums[z_idx] == target:
                    ans.append([nums[x_idx],nums[y_idx],nums[z_idx]])
        return ans
```

***这里存在一个非常关键的误区，就是双指针为什么复杂度是n^2的问题，可以理解为第二层遍历y_idx的时候里面再弄了一个while本质和for是并行的，就是两个指针互相移动的关系，所以合起来就是n的复杂度***
