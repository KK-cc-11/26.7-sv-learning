# leetcode 第一题的暴力解法 两层for O(n^2)的时间复杂度
```python 
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        l = len(nums)
        for i in range(l):
            for j in range(i + 1, l):
                if nums[i] + nums[j] == target:
                    return [i, j]  
```
**这是一种非常笨拙的方法，当我们想要查找这种对应关系的时候使用字典的数据结构是非常方便的**

# 题解提供的方法非常有趣
```python
class Solution:
    def twoSum(self, num: List[int], target: int) -> List[int]:
        record = dict()
        #创建一个字典用于记录for循环便历过的元素，键：num中的元素，值：每个元素的下标
        for idx, val in enumerate(nums):
            if (target - val) not in record:
                record[val] = idx
            else:
                return [target - val, idx]
```

# 一些反思
之所以题目采用字典的方式来处理数据，其目的是为了把时间复杂度下降到**On**级别
其实便历两次的目的就是为了拿一个**及时产生的**数据去对照
但是这产生的问题在于不利用一个存储空间用元素的value来查询对应idx是不明智的
我们可以用这种哈希map来代替一层for循环可以通过目标查找元素来解决On的问题
